### 1. Lagrange Interpolation

Let $(x_0, y_0), (x_1, y_1), (x_2, y_2), ..., (x_n, y_n)$ be a set of n+1 points, and $f(x)$ be the function to be interpolated. The value of $y$ corresponding to a given $x_i$ is set by $y_i = f(x_i)$. The form of $f(x)$ is not known. The Lagrange interpolation polynomial $L(x)$ is the polynomial of lowest degree that interpolates the given set of points. This is achieved by creating a polynomial basis using the given set of points.

The basis polynomials are given as follows:
$$l_i(x) = \left(\frac{x - x_0}{x_i - x_0}\right) \left(\frac{x - x_1}{x_i - x_1}\right) ... \left(\frac{x - x_{i-1}}{x_i - x_{i-1}}\right) \left(\frac{x - x_{i+1}}{x_i - x_{i+1}}\right) ... \left(\frac{x - x_n}{x_i - x_n}\right)$$
Note that the basis polynomial with the $i^{th}$ index excludes the $i^{th}$ point.

Given the basis polynomials, the Lagrange interpolation polynomial $L(x)$ of a degree $\leq n$ is given by:
$$L(x) = \sum_{i = 0}^{n} y_i l_i(x)$$



```python
def l_basis(x, x_vals, j):
    """
    Basis polynomial for Lagrange interpolation
        params:
        x: x value of a point
        x_vals: list of x values of all the points
        j: Index of the basis polynomial
    """
    basis = 1
    N = len(x_vals)
    for i in range(N):
        if j != i:
            basis = basis * (x - x_vals[i])/(x_vals[j] - x_vals[i])
    return basis

def lagrange(x, x_vals, y_vals):
    """
    Lagrange Interpolation of given lists of points x_vals and y_vals
    """
    N = len(x_vals)
    result = 0
    for i in range(N):
        result = result + l_basis(x, x_vals, i) * y_vals[i]
    return result
```

### 2. Cubic Spline Interpolation

Refer and rewrite: https://math.libretexts.org/Bookshelves/Applied_Mathematics/Numerical_Methods_(Chasnov)/05%3A_Interpolation/5.03%3A_Cubic_Spline_Interpolation

In the case of Cubic Spline Interpolation, the basis polynomial is a cubic polynomial. The interpolation is given by a piecewise function of all basis polynomials corresponding to a pair of consecutive points.
#### Definition
Let $(x_0, y_0), (x_1, y_1), (x_2, y_2), ..., (x_{n}, y_{n})$ be a set of $n+1$ points in strictly increasing order of $x_i$. Let $s_i(x)$ be the $i^{th}$ basis polynomial for the points in the interval $x_i \leq x < x_{i+1}$ for $i=0,1,2,..., n-1$. The basis is given by:
$$s_i(x) = a_i(x-x_i)^3 + b_i(x-x_i)^2 + c_i(x-x_i) + d_i$$
Then the cubic spline interpolation is given by the following piecewise function:
$$
    S(x) = 
    \begin{cases}
        s_0(x) & x_0 \leq x < x_1 \\
        s_1(x) & x_1 \leq x < x_2 \\
        s_2(x) & x_2 \leq x < x_3 \\
        ... \\
        s_{n-1}(x) & x_{n-1} \leq x < x_n \\
    \end{cases}
$$

#### Derivation
Since there are $n$ points, there are $n-1$ such equations for each cubic basis.
Taking the first and second derivatives of $s_i(x)$, we get the following:
$$\frac{ds_i}{dx} = 3a_i(x-x_i)^2 + 2b_i(x-x_i) + c_i \\

\frac{d^2s_i}{dx^2} = 6a_i(x-x_i) + 2b_i$$

We enforce the cubic spline interpolation $S(x)$ to obey the following properties:
1. $S(x)$ is continuous in $x\in [x_0, x_n]$
2. $S'(x)$ is continuous in $x\in [x_0, x_n]$
3. $S''(x)$ is continuous in $x\in [x_0, x_n]$

The continuity conditions can be stated as:
$$
\begin{align}
s_i(x_i) = y_i \text{ , } & i = 0,1,...,n-1 \\
s_i(x_{i+1}) = y_{i+1} \text{ , } & i = 0,1,...,n-1 \\
s'_i(x_{i+1}) = s'_{i+1}(x_{i+1}) \text{ , } & i = 0,1,...,n-2 \\
s''_i(x_{i+1}) = s''_{i+1}(x_{i+1}) \text{ , } & i = 0,1,...,n-2
\end{align}
$$

For the cubic spline to be continuous throughout the interval $[x_0, x_n]$, $S(x_i) = y_i$ must be true. Hence, it can be concluded that 
$$ 
a_i(x_i-x_i)^3 + b_i(x_i-x_i)^2 + c_i(x_i-x_i) + d_i \\
d_i = y_i \text{ where } i = 0, 1, 2,..., n-1
$$
Also, $s_i(x_i) = s_{i+1}(x_i)$ must also be held true if $S(x)$ is continuous. Using these relations, we obtain an equation for $d_i$:
$$
d_{i+1} = a_{i}h_i^3 +b_{i}h_i^2+c_{i}h_i+d_{i} \\
f_i = a_{i}h_i^3 +b_{i}h_i^2+c_{i}h_i \text{ , } i = 0, 1, 2,..., n-1
$$
Where $h_i = x_{i+1} - x_{i}$ and $f_i = y_{i+1} - y_i$

Using similar arguments for the first derivative of $S(x)$, we get the following equations for $c_i$ and $b_i$.

$$
c_{i+1} = 3a_{i} h_i^2 + 2b_{i}h_i + c_{i} \text{, } i = 0,1,2,...,n-2 \\
b_{i+1} = 3a_i h_i +b_i \text{, } i = 0,1,2,...,n-2
$$
By extending the above expression, $b_n = 3a_{n-1} h_{n-1} +b_{n-1}$.

By writing other coefficients in terms of $b_i$ and the known quantities, we get:
$$
    a_i = \frac{1}{3h_i} (b_{i+1} - b{i}) \text{, } i = 0,1,2,...,n-1 \\
    c_i = \frac{f_i}{h_i} - \frac{h_i}{3} (b_{i+1} + 2b_i) \text{, } i = 0,1,2,...,n-1 \\
$$
Substituting these values in the equation for $c_i$, we get:
$$
    \frac{1}{3} h_i b_i + \frac{2}{3}(h_i + h_{i+1}) b_{i+1} + \frac{1}{3} h_{i+1} b_{i+2} = \frac{f_{i+1}}{h_{i+1}} - \frac{f_i}{h_i} \text{, } i = 0,1,2,...,n-2 \\
$$
In matrix form, 
$$
    \begin{bmatrix}
        \frac{1}{3} h_0 & \frac{2}{3}(h_0 + h_{1}) & \frac{1}{3} h_{1} & 0 & ... & 0 & 0 \\
        0 & \frac{1}{3} h_1 & \frac{2}{3}(h_1 + h_{2}) & \frac{1}{3} h_{2} & ... & 0 & 0 \\
        \vdots & \vdots & \vdots & \vdots & \vdots & \vdots & \vdots \\
        0 & 0 & ... & \frac{1}{3} h_{n-3} & \frac{2}{3}(h_{n-3} + h_{n-2}) & \frac{1}{3} h_{n-2} & 0 \\
        0 & 0 & 0 & ... & \frac{1}{3} h_{n-2} & \frac{2}{3}(h_{n-2} + h_{n-1}) & \frac{1}{3} h_{n-1}
    \end{bmatrix}
    \begin{pmatrix}
        b_0 \\
        b_1 \\
        b_2 \\
        \vdots \\
        b_{n-2} \\
        b_{n-1} \\
        b_{n}
    \end{pmatrix}
    = 
    \begin{pmatrix}
        \frac{f_{1}}{h_{1}} - \frac{f_0}{h_0} \\
        \frac{f_{2}}{h_2} - \frac{f_1}{h_1} \\
        \vdots \\
        \frac{f_{n-1}}{h_{n-1}} - \frac{f_{n-2}}{h_{n-2}}
    \end{pmatrix}
$$
There are $n+1$ unknowns and $n-1$ equations.

For example, in matrix form for $n+1=5$, $n=4$, points $x_0$ to $x_4$, and with indices ($i = 0, 1, 2$) in the above equation:

$$
    \begin{bmatrix}
        \frac{1}{3} h_0 & \frac{2}{3}(h_0 + h_{1}) & \frac{1}{3} h_{1} & 0 & 0 \\
        0 & \frac{1}{3} h_1 & \frac{2}{3}(h_1 + h_{2}) & \frac{1}{3} h_{2} & 0 \\
        0 & 0 & \frac{1}{3} h_2 & \frac{2}{3}(h_2 + h_{3}) & \frac{1}{3} h_{3}
    \end{bmatrix}
    \begin{pmatrix}
        b_0 \\
        b_1 \\
        b_2 \\
        b_3 \\
        b_4
    \end{pmatrix}
    = 
    \begin{pmatrix}
        \frac{f_{1}}{h_{1}} - \frac{f_0}{h_0} \\
        \frac{f_{2}}{h_2} - \frac{f_1}{h_1} \\
        \frac{f_{3}}{h_3} - \frac{f_2}{h_2}
    \end{pmatrix}
$$


System is under-determined with $n+1=5$ variables and $n-1=3$ equations.

#### Choice of second derivative to eliminate two variables for a determined system
#### Natural spline:
The **natural choice** of cubic spline sets the 2nd derivative at the endpoints, i.e., $b_0$ and $b_n$, to be zero. The first and last columns can be removed:
$$
    \begin{bmatrix}
         \frac{2}{3}(h_0 + h_{1}) & \frac{1}{3} h_{1} & 0 & ... & 0 \\
        \frac{1}{3} h_1 & \frac{2}{3}(h_1 + h_{2}) & \frac{1}{3} h_{2} & ... & 0 \\
         \vdots & \vdots & \vdots & \vdots & \vdots \\
        0 & ... & \frac{1}{3} h_{n-3} & \frac{2}{3}(h_{n-3} + h_{n-2}) & \frac{1}{3} h_{n-2} \\
        0 & 0 & ... & \frac{1}{3} h_{n-2} & \frac{2}{3}(h_{n-2} + h_{n-1})
    \end{bmatrix}
    \begin{pmatrix}
        b_1 \\
        b_2 \\
        \vdots \\
        b_{n-2} \\
        b_{n-1} \\
    \end{pmatrix}
    = 
    \begin{pmatrix}
        \frac{f_{1}}{h_{1}} - \frac{f_0}{h_0} \\
        \frac{f_{2}}{h_2} - \frac{f_1}{h_1} \\
        \vdots \\
        \frac{f_{n-1}}{h_{n-1}} - \frac{f_{n-2}}{h_{n-2}}
    \end{pmatrix}
$$
In the above system, the choice of second derivative can change the matrix into the following by removing the first and last columns: 
$$
    \begin{bmatrix}
        \frac{2}{3}(h_0 + h_{1}) & \frac{1}{3} h_{1} & 0  \\
        \frac{1}{3} h_1 & \frac{2}{3}(h_1 + h_{2}) & \frac{1}{3} h_{2}  \\
        0 & \frac{1}{3} h_2 & \frac{2}{3}(h_2 + h_{3})
    \end{bmatrix}
    \begin{pmatrix}
        b_1 \\
        b_2 \\
        b_3 \\
    \end{pmatrix}
    = 
    \begin{pmatrix}
        \frac{f_{1}}{h_{1}} - \frac{f_0}{h_0} \\
        \frac{f_{2}}{h_2} - \frac{f_1}{h_1} \\
        \frac{f_{3}}{h_3} - \frac{f_2}{h_2}
    \end{pmatrix}
$$
#### Parabolic runout spline:
The **parabolic choice** of cubic spline sets $b_0 = b_1$ and $b_n= b_{n-1}$. 
$$
    \begin{bmatrix}
         (h_0 + \frac{2}{3} h_{1}) & \frac{1}{3} h_{1} & 0 & ... & 0 \\
        \frac{1}{3} h_1 & \frac{2}{3}(h_1 + h_{2}) & \frac{1}{3} h_{2} & ... & 0 \\
         \vdots & \vdots & \vdots & \vdots & \vdots \\
        0 & ... & \frac{1}{3} h_{n-3} & \frac{2}{3}(h_{n-3} + h_{n-2}) & \frac{1}{3} h_{n-2} \\
        0 & 0 & ... & \frac{1}{3} h_{n-2} & (\frac{2}{3} h_{n-2} + h_{n-1})
    \end{bmatrix}
    \begin{pmatrix}
        b_1 \\
        b_2 \\
        \vdots \\
        b_{n-2} \\
        b_{n-1} \\
    \end{pmatrix}
    = 
    \begin{pmatrix}
        \frac{f_{1}}{h_{1}} - \frac{f_0}{h_0} \\
        \frac{f_{2}}{h_2} - \frac{f_1}{h_1} \\
        \vdots \\
        \frac{f_{n-1}}{h_{n-1}} - \frac{f_{n-2}}{h_{n-2}}
    \end{pmatrix}
$$

For $n = 4$, the system becomes:
$$
    \begin{bmatrix}
        (h_0 + \frac{2}{3}h_{1}) & \frac{1}{3} h_{1} & 0  \\
        \frac{1}{3} h_1 & \frac{2}{3}(h_1 + h_{2}) & \frac{1}{3} h_{2}  \\
        0 & \frac{1}{3} h_2 & (\frac{2}{3}h_2 + h_{3})
    \end{bmatrix}
    \begin{pmatrix}
        b_1 \\
        b_2 \\
        b_3 \\
    \end{pmatrix}
    = 
    \begin{pmatrix}
        \frac{f_{1}}{h_{1}} - \frac{f_0}{h_0} \\
        \frac{f_{2}}{h_2} - \frac{f_1}{h_1} \\
        \frac{f_{3}}{h_3} - \frac{f_2}{h_2}
    \end{pmatrix}
$$
